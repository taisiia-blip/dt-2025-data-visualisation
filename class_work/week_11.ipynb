```python
sentence = "aquickbrownfoxjumpedoverthelazydog" #34 letters

#  extract first letter
print("Print first letter", sentence[0:1]) #start:end
print(sentence[:3])
# exctrct lettres from  5 to 9
print("Print letters from 5-9:", sentence[5:9])
# exctract letters from 5t to end
print(sentence[5:])
# Extract the lastcharater 
print(sentence[-1])
#Exract last 4 characters
print("Last four", sentence[-4:])
#extract first letter of the ubstring startingfrom 5
print(sentence[5:][0]) #sentence[5:] extracts a substring starting from 5; [0]: thi will take the first character of that substring
#Extract first 5 letters of the substring starting from 5
print(sentence [5:][:5])
print(sentence[:]) #extract ntire string

# extract every 2nd character
print(sentence[::2])
```

```python
Sentence1 = "Alice went to the park and saw beautiful bird singing"
words = Sentence1.split(" ") # splits te sentence to words
print(words)
#Extract only the first word
print(words[0])
#extract the first 3 words
print(*words[0:3]) # to remove brackets
#extract 3 last words
print(*words[-3:])
#print number of words
print("Number of words", len(words))
```
