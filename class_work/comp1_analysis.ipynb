```python
import pandas as pd
df = pd.read_csv("comp1.csv")
df 
```

```python
import matplotlib.pyplot as plt
avg_hours=df.groupby("Student") ["Hours_Studied"].mean()
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied", legend= False)
plt.title("Who_puts_in_most_efforts")
plt.ylabel("Hours")
plt.show()

```

```python
avg_grade=df.groupby("Student")["Grade"].mean().sort_values(ascending= False)
avg_grade.plot(kind="bar", x="Student", y="Grade")
plt.title("Who performs best?")
plt.ylabel("Grades")
plt.show()
```

```python
avg_grade=df.groupby("Student")["Grade"].mean()
avg_grade.plot(kind="pie", x="Student", y="Grade")
plt.title("Who performs best?")
plt.ylabel("Grades")
plt.show()
```

```python

avg_hours=df.groupby("Student") ["Hours_Studied"].mean()
plt.figure(figsize=(8,8))
avg_hours.plot(kind="pie", x="Student", y="Hours_Studied", legend= False, autopct= '%1.1f%%')
plt.title("Who_puts_in_most_efforts")
plt.ylabel("Hours")
plt.show()
```

Does more studying = better grades?

```python
plt.scatter(df["Hours_Studied"], df["Grade"])
plt.show()
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.title("Does more studying = better grades?")
plt.xlabel("Hours studied")
plt.ylabel("Grades")
plt.show()
```

How consistent are study patterns and performances?

```python
# Calculate quartiles and ekstremes
hours_min = df["Hours_Studied"].min()
hours_q1=df["Hours_Studied"].quantile(0.25)
print("Minimun hours:", hours_min, "Q1:", hours_q1)


plt.figure(figsize=(8,6))
df[["Hours_Studied","Grade"]].boxplot()
plt.title("How consistent are study patters and performance?")
plt.ylabel("Value")
plt.show()
```
