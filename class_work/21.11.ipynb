Simple code to make a word cloud

```python
from wordcloud import WordCloud
wc = WordCloud().generate("rain brings hope hope big bunnies big big kittens")
img = wc.to_image()
img.show()
# "big" is the bigger size based on its frequency 
```

```python
import urllib.request # this we use to work with URLs

text = urllib.request.urlopen("https://raw.githubusercontent.com/taisiia-blip/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/class_work/yo.txt").read().decode("utf-8") # decode("utf-8") to convert it into a readible file for python
text[:50]

WordCloud().generate_from_text(text).to_image()

```

```python
WordCloud(stopwords=["and", "then", "but", "the", "to", "on", "of"]).generate_from_text(text).to_image() # omits the words that dont give us any meaning

```

```python
WordCloud(stopwords=["and", "then", "but", "the", "to", "on", "of"],background_color="white").generate_from_text(text).to_image() # changes the background color to white
```

```python
#in order to work with image masks, we need numpy
from PIL import Image #impot image class from PIL to work wit images
import urllib.request #library to work with URLs 
import numpy as np 
text3 = urllib.request.urlopen("https://drive.google.com/uc?export=download&id=1WBQ4_MURlW8Qnj0eWakwI6W4FYRllfX_").read().decode("utf-8")
img = urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1we4oHVBnX7bKecDKzSj_YTvdR3L0xY50","image1.png")

WordCloud().generate_from_text(text3).to_image() # norml
WordCloud(mask=np.array(Image.open("image1.png"))).generate_from_text(text3).to_image() # to mask it to be in a circle
WordCloud(mask=np.array(Image.open("image1.png")),background_color="white").generate_from_text(text3).to_image() # change the bg color
```
