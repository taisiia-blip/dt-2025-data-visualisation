```python

import urllib.request
luuletus = urllib.request.urlopen("https://raw.githubusercontent.com/taisiia-blip/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/class_work/assignment/karlristikivi.txt").read().decode("utf-8") # reads and decodes the text in a readable format for python
# trying out something new: to clean up the text i will use the regular expressions like in notebook++, but in python. I found this out when consulting with AI about the ways to remove the newlines
import re # makes it possible to use a reg expression

clean = luuletus.replace("\n", " ") # replaces the newlines with blank spaces
words = re.split(r"[ ,\.]+", clean) # takes the clean text and splits it into separate word, by comma and period. 
#the +´indicates that there are more than 1 separator. # r means that the backslashes stay literal, not interpreted as code
print(words)
print(words[0]) # prints the first word
print(words[-9:]) # prints the last 9 words of the poem
print("Number of words", len(words)) # prints the total number of words


```

```python
from wordcloud import WordCloud
import urllib.request
luuletus = urllib.request.urlopen("https://raw.githubusercontent.com/taisiia-blip/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/class_work/assignment/karlristikivi.txt").read().decode("utf-8")
luuletus[:101]
WordCloud(stopwords=["ja", "et", "on", "aga", "nõnda", "või", "nii", "veel", "kui", "kuigi", "ka", "nagu"]).generate_from_text(luuletus).to_image() # generates an image omitting some words
```

This wordcloud shows that the dominant words in this poem are "ma" and "ei" which mean "I" and "no" consequtively. These word are followed after: "vesi"(water) and "Arkaadia (tee)" (The road of Arcadia), as well as "kaotasin"(I lost). Based on this word cloud, the authors message says that he is talking about his own perspective, about loss and a long journey.

```python
WordCloud(stopwords=["ja", "et", "on", "aga"], background_color="white", max_words=25).generate_from_text(luuletus).to_image() # the background color is changed to white, the maximum amount of words is 25
```
