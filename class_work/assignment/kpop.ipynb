```python
import pandas as pd
df= pd.read_csv("kpop.csv")
df
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="group_name", y= "music_show_awards", legend= False)
plt.ylabel("Music Awards")
plt.title("Music awards per each group")
plt.show
```

In this chart, we see every music awards won by the given group. The most awards were won by the group "Girl's Generation" while the least were won by "TREASURE".

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9,6))
df.plot(kind="pie", x="group_name", y="physical_sales", legend= False, autopct= '%1.1f%%')
plt.ylabel("Sales")
plt.title("Sales per group")
```

This chart shows the numbers of physical sales for each group. Group number 0, ITZY, is shown to have the biggest amount of sales.

```python
plt.scatter(df["music_show_awards"], df["physical_sales"])
plt.title("Does more award = better sales?")
plt.xlabel("Music show awards")
plt.ylabel("Sales")
plt.show()
```

This charts shows correlations between music show awards and sales. In here, we can see a general corrlation that the less awards the groups have, the lesser are the sales. But on the other hand, we can see outliers such as high sales with low number of rewards (6 thousand sales, <20 awards) and low sales with high number of awards (4< thousand sales, <80 rewards)

```python
awards_min= df["music_show_awards"].min()
awards_q1= df["music_show_awards"].quantile(0.25)
awards_q3= df ["music_show_awards"].quantile(0.75)
print("Minimum awards:", awards_min, "Q1:", awards_q1, "Q3:", awards_q3)

plt.figure(figsize=(8,6))
df[["music_show_awards", "physical_sales"]].boxplot()
plt.title("How consistent are sales and awards")
plt.ylabel("Value")
plt.show()
```

This chart shows the difference between the number of sales compared to awards, as well as the extremes and quartiles of the given data.
