```python
import pandas as pd 

df = pd.read_csv("sports.csv")
df
```

```python
df.describe() # generate some statistics
```

box plot for each athelete to find the number of years of practice  

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Years_of_practice", legend= False)
plt.ylabel("Years")
plt.title("Years of practice per athlete")
plt.show
```

```python
df.plot(kind="bar", x="Name", y= "Age", legend=False)
plt.ylabel("Age")
plt.title("Ages of athletes")
plt.show
```

plt.show
"s" - controls the size

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*30, alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Years")
plt.title("Age vs years of practice")
plt.show
# *30 means that the size of the bubble will accord to the number, *30 is a scaling factor to make the bubbles big
#  alpha=0.6 means it will be transparent if it overlaps with other bubbles
```

Average age per sport

```python
avg_age=df.groupby("Sport") ["Age"].mean()
avg_age.plot(kind="bar")
plt.title("Average age per sport")
plt.ylabel("Age")
plt.show()
```
